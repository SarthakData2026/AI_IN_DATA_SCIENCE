#line plot
-show trend over continuous data
-time series

```python
import matplotlib.pyplot as plt
```

```python
marks=[98,90,55,56]
stud_hr=[10,9,5,6]
```

```python
plt.plot(marks,stud_hr,marker="s")
for i in range(len(marks)):
  plt.text(marks[i],stud_hr[i],str([marks[i],stud_hr[i]]),ha="center")
```

```python
plt.title("student status")
```

```python
plt.xlabel("marks")
plt.ylabel("stud_hr")
```

```python
plt.figure(figsize=(10,10))
```

```python
plt.plot(marks,stud_hr,marker='s',linestyle='--',color='red')
plt.xlabel("marks")
plt.ylabel("stud_hr")
plt.title("student status",fontsize=16, color='green')
plt.grid(True)
```

```python
plt.show()
```

#bar chart
-category

```python
subj=["python","java","c","das"]
score=[20,30,40,50]
```

```python
plt.bar(subj,score,color=["green","blue","red","yellow"])
plt.title("score of students",fontsize=16)
plt.xlabel("subject")
plt.ylabel("score")
plt.grid(axis="y",linestyle="--")
plt.show()
```

#3.4 pie chart

```python
sub=["tech","eng","apti","gk"]
score=[30,40,50,60]
colours=["green","blue","red","yellow"]
plt.pie(score,labels=sub,colors=["green","blue","red","yellow"],autopct="%1.1f%%",startangle=180,explode=(1,0,0,0))
plt.title("score of students",fontsize=16)
```

```python
sub=["tech","eng","apti","gk"]
score=[30,40,50,60]
colours=["green","blue","red","yellow"]
plt.pie(score,labels=sub,colors=["green","blue","red","yellow"],autopct="%1.1f%%",startangle=180,explode=(1,0,0,0),shadow=True)
plt.legend(loc="best",fontsize=8)

plt.title("score of students",fontsize=16)
```

```python
x=[2,4,6,100,8]
y=[1,3,5,7,50]
rang=['red','blue','green','yellow','pink']
plt.scatter(x,y,c=rang)
plt.title("find outliar ")
plt.xlabel("first ")
plt.ylabel("second")
plt.colorbar(label='intensity')
```

```python
x=[2,4,6,100,8]
y=[1,3,5,7,50]
Size=[100,200,300,400,500]
rang=['red','blue','green','yellow','pink']
plt.scatter(x,y,c=rang,s=Size)
plt.title("find outliar ")
plt.xlabel("first ")
plt.ylabel("second")
plt.colorbar(label='intensity')
```
