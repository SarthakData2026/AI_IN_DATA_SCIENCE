```python
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd
import numpy as np
```

```python
#import data set
df=px.data.gapminder()
```

```python
canada_df=px.data.gapminder().query("country=='Canada'")
```

```python
canada_df
```

#bar chart

```python
fig=px.bar( canada_df,x='year',y='pop' ,title="canada population status")
```

```python
fig=px.bar( canada_df,x='year',y='pop' ,title="canada population status",labels={'pop':'population'})
```

```python
fig=px.bar( canada_df,x='year',y='pop' ,title="canada population status",labels={'pop':'population'},color='pop')
```

```python
fig.show()
```

```python
fig=px.bar( canada_df,x='year',y='pop' ,title="canada population status",labels={'pop':'population'},color='pop',hover_data=['lifeExp'],text='pop')
```

```python
fig.show()
```

```python
#customize text show in bar
fig.update_traces(texttemplate='%{text:.2s}',textposition='outside')
fig.show()
```

```python
#update layout
fig.update_layout(xaxis_tickangle=-45)
fig.show()
```

```python
#update layout
fig.update_layout(xaxis_tickangle=-45,yaxis_title='population of canada')
fig.show()
```

```python
from tempfile import template
#update layout
fig.update_layout(xaxis_tickangle=-45,yaxis_title='population of canada',template='plotly_dark')
fig.show()
```

```python
from tempfile import template
#update layout
fig.update_layout(xaxis_tickangle=-45,yaxis_title='population of canada',template='plotly_dark',uniformtext_minsize=10)
fig.show()
```

```python
#update layout
fig.update_layout(xaxis_tickangle=-45,yaxis_title='population of canada',template='plotly_dark',uniformtext_minsize=10,uniformtext_mode='hide')
fig.show()
```

```python
#update layout
fig=px.line( canada_df,x='year',y='pop' ,title="canada population status",labels={'pop':'population'},color='pop',hover_data=['lifeExp'],text='pop')
fig.update_layout(xaxis_tickangle=-45,yaxis_title='population of canada',template='plotly_dark',uniformtext_minsize=10,uniformtext_mode='hide')
fig.show()
```
