#join

```python
v1=pd.DataFrame({"rollnum":[10,20,30],"name":["sarthak","dev","pragya"]})
v2=pd.DataFrame({"rollnum1":[10,20,30],"name1":["a","b","c"]})
v1.join(v2,how="left")
```
